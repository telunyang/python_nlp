# 隱藏式馬可夫模型：咖啡廳對話

NPC 說話很友善，是否就代表喜歡你？這個虛構小遊戲用 HMM 示範：**同一種反應，可以來自不同心情**。

本課保留四個重點：隱藏狀態與觀測、轉移與發射機率、Viterbi 路徑、後驗機率。
先當「導演」編寫三場對話，再當「玩家」只看反應推理。

這是遊戲角色的示意模型，不是用來判定真人感情的工具。

```python
# 名稱順序也就是矩陣的列、欄順序。
import numpy as np
import pandas as pd
from hmmlearn.hmm import CategoricalHMM
from IPython.display import display

states = ["不喜歡", "有好感", "喜歡"]
reactions = ["愉快地互動", "友善回應", "冷漠"]
```

## 1. 導演模式：用對話紀錄取代手填機率

每個 tuple 依序是 **（NPC 內心、可見反應、NPC 臺詞）**。一個內層清單就是一場獨立的對話。
有空的話，可以讓同學扮演 NPC，演完幾回合後由「導演」補上內心標籤，再修改這份紀錄。

內心是遊戲劇本的已知設定，玩家推論時看不到；反應類別也由人標記，程式不會自動理解臺詞。
本例使用**有狀態標籤（state）的紀錄做計數估計**，不是只靠幾句文字就自動學會 HMM。
如果只有反應而沒有狀態標籤，則是另一個學習問題，可進一步學習 Baum–Welch。

```python
# 修改或增加對話後，重新執行後面的格子，矩陣就會跟著改變。
scenes = [
    [
        ("不喜歡", "冷漠", "我現在不太想聊天。"),
        ("不喜歡", "友善回應", "謝謝你幫忙拿咖啡。"),
        ("有好感", "友善回應", "原來你也常來這間店？"),
        ("喜歡", "愉快地互動", "跟你聊天真開心，下次再一起來！"),
    ],
    [
        ("有好感", "友善回應", "嗨，今天過得如何？"),
        ("喜歡", "愉快地互動", "你居然記得我喜歡的口味！"),
        ("喜歡", "友善回應", "謝謝，我很喜歡這杯咖啡。"),
        ("喜歡", "愉快地互動", "下次換我請你！"),
    ],
    [
        ("有好感", "冷漠", "今天有點累，先讓我坐一下。"),
        ("不喜歡", "冷漠", "這個玩笑讓我不太舒服。"),
        ("有好感", "友善回應", "謝謝你道歉，我們換個話題吧。"),
        ("有好感", "愉快地互動", "哈哈，你家的貓也太可愛了！"),
    ],
]

# 讀出小劇場；括號中的內心是導演筆記，不是模型的預測。
for number, scene in enumerate(scenes, start=1):
    print(f"\n第 {number} 場")
    for state, reaction, line in scene:
        print(f"NPC：{line}　【反應：{reaction}；內心：{state}】")
```

## 2. 先數次數，暫時不要想機率

| 要數什麼 | 怎麼數 | 用途 |
|---|---|---|
| 開場狀態 | 每場只看第一回合的內心 | 初始機率 |
| 心情轉移 | 同場的「上一回合內心 → 這一回合內心」 | 轉移機率矩陣 |
| 外在反應 | 每回合的「當下內心 → 當下反應」 | 發射機率矩陣 |

例如第一場的心情是「不喜歡 → 不喜歡 → 有好感 → 喜歡」，所以有三次轉移。
**不同場次不相連**：第一場的結尾不能算成轉移到第二場的開頭。
發射機率描述的是反應在某種心情下出現的機會，不是反應造成心情改變的機率。

```python
# 先建立全為 0 的計數表；用中文名稱定位，避免混淆數字索引。
start_counts = pd.Series(0, index=states, name="開場次數")
transition_counts = pd.DataFrame(0, index=states, columns=states)
emission_counts = pd.DataFrame(0, index=states, columns=reactions)

# 計算每場的開場、轉移與發射次數
for scene in scenes:
    # 計算每場的開場次數
    start_counts[scene[0][0]] += 1

    # 計算每場的轉移與發射次數
    previous_state = None

    # 逐行處理每場對話
    for state, reaction, line in scene:
        # 計算發射次數
        emission_counts.loc[state, reaction] += 1

        # 計算轉移次數
        if previous_state is not None:
            # 如果 previous_state 不是 None，表示這不是開場行，才計算轉移次數
            transition_counts.loc[previous_state, state] += 1
        
        # 更新 previous_state 為當前的 state，以便下一次迭代使用
        previous_state = state

display(start_counts.to_frame())

print("轉移次數：列＝目前心情，欄＝下一回合心情")
display(transition_counts)

print("發射次數：列＝當下心情，欄＝當下反應")
display(emission_counts)
```

## 3. 次數除以總次數，就是機率

每一列都代表一個條件，因此**每一列的機率總和必須是 1**。
資料只有 12 回合，沒看過的組合不代表永遠不可能；這裡採用「每格加 1」的平滑方式：

$$P = \frac{\text{該格次數}+1}{\text{該列總次數}+\text{該列欄位數}}$$

以原始劇本為例，「不喜歡」的轉移次數為 `[1, 2, 0]`，加 1 後是 `[2, 3, 1]`，
除以 6 得到 `[0.333, 0.500, 0.167]`。現在每個數字都能回頭找到對話紀錄。
平滑也會影響結果；少量對話只能示範計算方式，不能代表穩定估計。

```python
# 兩個矩陣共用相同的「加 1，再逐列正規化」步驟。
def row_probabilities(counts):
    """把每一列的計數轉成總和為 1 的機率。"""
    # 加 1 平滑，避免出現 0 機率。
    smoothed = counts + 1

    # 逐列正規化，讓每一列的總和為 1。
    return smoothed.div(smoothed.sum(axis=1), axis=0)

# 計算初始機率、轉移機率與發射機率
start_prob = (start_counts + 1) / (start_counts.sum() + len(states))

# 轉移機率與發射機率的計算，使用 row_probabilities 函數。
transition_prob = row_probabilities(transition_counts)

# 計算發射機率
emission_prob = row_probabilities(emission_counts)

# 只將顯示值四捨五入，模型仍使用完整精度。
display(start_prob.to_frame("初始機率").round(3))

print("轉移機率：列＝目前心情，欄＝下一回合心情")
display(transition_prob.round(3))

print("發射機率：列＝當下心情，欄＝當下反應")
display(emission_prob.round(3))
```

## 4. 把剛算好的機率交給 HMM

`startprob_`、`transmat_`、`emissionprob_` 就是上面的三份機率。
`init_params=""` 表示不使用自動初始化；本課已從劇本計數得到參數，所以直接推論即可。
HMM 假設下一回合的狀態只依賴當前狀態，而當前反應只依賴當前狀態。
這個基本模型沒有另外納入玩家選項、說話內容或其他事件對狀態的影響。

```python
# 直接使用計數估計的參數，不額外呼叫 fit()。
model = CategoricalHMM(
    n_components=len(states), # 狀態數量
    n_features=len(reactions),  # 觀察值數量
    init_params="" # 不額外初始化參數，直接使用我們計算的機率
)

# 將計算好的機率轉換為 NumPy 陣列，並指定給模型的屬性
model.startprob_ = start_prob.to_numpy()

# 將計算好的轉移機率與發射機率轉換為 NumPy 陣列，並指定給模型的屬性
model.transmat_ = transition_prob.to_numpy()

# 將計算好的發射機率轉換為 NumPy 陣列，並指定給模型的屬性
model.emissionprob_ = emission_prob.to_numpy()
```

## 5. 玩家模式：只看反應，推測一路上的心情

把下一格的 `observed` 當作你剛經歷的新對話：逐回合填入三種反應之一，再執行。
也可以每次只增加一個反應，觀察新的證據如何影響判斷。

- **Viterbi**：找出整條聯合機率最高的狀態路徑，不等於逐列挑最大的後驗機率。
- **後驗機率**：使用本次輸入的**整段反應**推測各回合的狀態，因此後面的反應也可能改變前面回合的判斷。

表格最後一列只使用截至目前的證據；其他列則是看完整段後的回顧。
這裡不另外設定「超過 0.8 就是真心」之類的門檻，直接觀察模型的不確定性。

```python
# 每次改寫或增加一個反應，就能重新進行這場推理遊戲。
observed = ["友善回應", "冷漠", "友善回應", "愉快地互動"]

# 檢查輸入的反應是否在已知的 reactions 列表中，若不在則拋出錯誤。
if not observed or any(reaction not in reactions for reaction in observed):
    # 如果 observed 為空或包含不在 reactions 中的反應，則拋出 ValueError。
    raise ValueError(f"請輸入至少一個反應，名稱限於：{reactions}")

# CategoricalHMM 接受 0 起算的類別編號，形狀為（回合數, 1）。
X = np.array([reactions.index(reaction) for reaction in observed]).reshape(-1, 1)

# 預測最可能的心情序列
path = model.predict(X)  # 預設使用 Viterbi 解碼。

# 計算每個回合的後驗機率
posterior = model.predict_proba(X)

# 用一張表取代額外的排序、敘事模板與判讀門檻。
result = pd.DataFrame(posterior, columns=states, index=range(1, len(observed) + 1))

# 顯示結果表格
result.index.name = "回合"
result.insert(0, "看到的反應", observed)
result.insert(1, "Viterbi 心情", [states[index] for index in path])
display(result.round(3))
```
