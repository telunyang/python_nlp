# 嘗試簡單地分詞，以及去除停用詞

```python
# 輸入文本
text = "這是使用 Jieba 和 sklearn 進行中文字預處理的範例。"

# 分詞 (cut 是用於 iteration 的斷詞函式，lcut 是將斷詞以 list 格式回傳)
tokens = jieba.lcut(text)

# 去除停用詞，這邊為了簡化，我們預先定義一個停用詞列表
stop_words = ["和", "的", "是", "這"]
tokens = [word for word in tokens if word not in stop_words]

# 顯示過濾後的斷詞結果。
tokens
```

```python
# POS tagging
import jieba.posseg as pseg

# 假設我們有以下句子
sentence = "我喜歡看電影"

# 使用jieba進行詞性標註
words = pseg.cut(sentence)

# 列印每個詞及其詞性
for word, flag in words:
    print(f'{word} {flag}')
```

## 從 Jieba 詞性走向依存分析
Jieba 的詞性標註不提供支配詞或依存樹。

本例使用容易觀察主語、動詞與地點的短句。模型能接受繁體中文，但不代表每個詞都分析正確；
請把斷詞、詞性與依存關係分別檢查，不要將模型預測視為標準答案。

## spaCy 依存句法分析與視覺化
依存分析回答「哪個詞修飾或依附哪個詞」，不同於只標示詞性的 POS tagging，也不同於 NP / VP 成分樹。
請保留原文的大小寫、停用詞與標點，讓模型使用完整語境。

表格的 `head` 是支配詞，`dep` 是此詞與支配詞的關係；`ROOT` 的 head 是自己。
箭頭由支配詞指向依存詞，ROOT 不會畫出指向自己的箭頭。
常見關係：`nsubj` 主語、`dobj` / `obj` 受詞、`amod` 形容詞修飾、`advmod` 副詞修飾。
不同語言模型的標籤不一定相同，請搭配表格中的解釋判讀。

參考：[spaCy 語言特徵](https://spacy.io/usage/linguistic-features)、[displaCy 視覺化](https://spacy.io/usage/visualizers)。

```python
# 模型只需載入一次；分析與視覺化使用中文。
import spacy
import pandas as pd
from spacy import displacy
from IPython.display import HTML, display

nlp = spacy.load("zh_core_web_sm")
source_text = "小明在學校讀書。老師在教室教學生。"
doc = nlp(source_text)
print("原文：", source_text)
print("模型：", nlp.meta["name"], "版本：", nlp.meta["version"])
```

```python
# 以索引區分重複出現的詞，並列出每個詞的支配詞與關係。
def dependency_table(doc):
    """將逐詞分析整理成便於閱讀的表格。"""
    return pd.DataFrame([
        {"i": token.i, "text": token.text, "pos": token.pos_,
         "tag": token.tag_, "dep": token.dep_,
         "head_i": token.head.i, "head": token.head.text,
         "說明": spacy.explain(token.dep_) or "請參考模型標籤定義"}
        for token in doc if not token.is_space
    ])

dependency_table(doc)
```

```python
# 每句各畫一張圖；長句可水平捲動，避免圖形被邊界裁切。
def show_dependencies(doc):
    """用 displaCy 顯示每句的詞性與依存箭頭。"""
    for sentence in doc.sents:
        print(sentence.text)
        svg = displacy.render(
            sentence, 
            style="dep", 
            options={
                "compact": False,
                "distance": 110,
                "font": "Microsoft JhengHei, Arial, sans-serif",
            }
        )
        display(HTML(f'<div style="overflow-x:auto; padding:16px">{svg}</div>'))

show_dependencies(doc)
```
